```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import sys
from pathlib import Path

# Aggiunge la root del progetto al PYTHONPATH
# In Jupyter notebooks, usiamo il working directory corrente e navighiamo alla root
current_dir = Path.cwd()
# Se siamo nella cartella notebooks, andiamo su di un livello
if current_dir.name == 'notebooks':
    project_root = current_dir.parent
else:
    project_root = current_dir

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

# Cambia directory alla root del progetto per i path relativi
os.chdir(project_root)

# Carica il file CSV dei gradienti
df_grad = pd.read_csv("logs/grad_stats_freezing.csv")

# Raggruppa per epoch e calcola la media dei gradienti per ogni epoch
df_grouped = df_grad.groupby('epoch').agg({
    'pose_grad_mean': 'mean',
    'pose_grad_std': 'mean',
    'text_grad_mean': 'mean',
    'text_grad_std': 'mean'
}).reset_index()

# Plot
plt.figure(figsize=(12, 6))
plt.plot(df_grouped['epoch'], df_grouped['pose_grad_mean'], label='Pose Encoder - Mean Gradients', linewidth=2)
plt.plot(df_grouped['epoch'], df_grouped['text_grad_mean'], label='Text Encoder - Mean Gradients', linewidth=2)
plt.fill_between(df_grouped['epoch'], 
                 df_grouped['pose_grad_mean'] - df_grouped['pose_grad_std'], 
                 df_grouped['pose_grad_mean'] + df_grouped['pose_grad_std'], 
                 alpha=0.2)
plt.fill_between(df_grouped['epoch'], 
                 df_grouped['text_grad_mean'] - df_grouped['text_grad_std'], 
                 df_grouped['text_grad_mean'] + df_grouped['text_grad_std'], 
                 alpha=0.2)

# Calcola min e max considerando anche le bande di deviazione standard
y_min = min(
    (df_grouped['pose_grad_mean'] - df_grouped['pose_grad_std']).min(),
    (df_grouped['text_grad_mean'] - df_grouped['text_grad_std']).min()
)
y_max = max(
    (df_grouped['pose_grad_mean'] + df_grouped['pose_grad_std']).max(),
    (df_grouped['text_grad_mean'] + df_grouped['text_grad_std']).max()
)

# Imposta i limiti dell'asse y con 0.1 sopra il max e 0.1 sotto il min
plt.ylim(y_min - 0.1, y_max + 0.1)

plt.title("Confronto Gradienti - Pose vs Text Encoder (Congelamento Parziale)")
plt.xlabel("Epoch")
plt.ylabel("Media Gradienti")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import sys
from pathlib import Path

# Aggiunge la root del progetto al PYTHONPATH
# In Jupyter notebooks, usiamo il working directory corrente e navighiamo alla root
current_dir = Path.cwd()
# Se siamo nella cartella notebooks, andiamo su di un livello
if current_dir.name == 'notebooks':
    project_root = current_dir.parent
else:
    project_root = current_dir

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

# Cambia directory alla root del progetto per i path relativi
os.chdir(project_root)

# Carica il file CSV dei gradienti
df_grad = pd.read_csv("logs/grad_stats_freezing_adapter.csv")

# Raggruppa per epoch e calcola la media dei gradienti per ogni epoch
df_grouped = df_grad.groupby('epoch').agg({
    'pose_grad_mean': 'mean',
    'pose_grad_std': 'mean',
    'text_grad_mean': 'mean',
    'text_grad_std': 'mean'
}).reset_index()

# Plot
plt.figure(figsize=(12, 6))
plt.plot(df_grouped['epoch'], df_grouped['pose_grad_mean'], label='Pose Encoder - Mean Gradients', linewidth=2)
plt.plot(df_grouped['epoch'], df_grouped['text_grad_mean'], label='Text Encoder - Mean Gradients', linewidth=2)
plt.fill_between(df_grouped['epoch'], 
                 df_grouped['pose_grad_mean'] - df_grouped['pose_grad_std'], 
                 df_grouped['pose_grad_mean'] + df_grouped['pose_grad_std'], 
                 alpha=0.2)
plt.fill_between(df_grouped['epoch'], 
                 df_grouped['text_grad_mean'] - df_grouped['text_grad_std'], 
                 df_grouped['text_grad_mean'] + df_grouped['text_grad_std'], 
                 alpha=0.2)
plt.title("Confronto Gradienti - Pose vs Text Encoder (Congelamento Parziale)")
plt.xlabel("Epoch")
plt.ylabel("Media Gradienti")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

Il grafico mostra chiaramente che, nonostante l’inserimento dell’Adapter Layer e il congelamento dei pesi BERT tranne l’ultimo layer:

Il pose encoder continua ad avere gradienti medi più alti durante tutto l’addestramento.

Il text encoder mantiene gradienti medi molto bassi, tra 0.2 e 0.5, indicando che anche l’ultimo layer + projection e adapter non stanno assorbendo sufficiente segnale di apprendimento.

Diagnosi aggiornata:

L’overfitting persiste: la val loss crolla fino all’epoca 10–11, poi risale.

I gradienti sono ancora squilibrati: il text encoder non apprende abbastanza rispetto al pose encoder, anche se l’architettura è più "flessibile" ora.

La bassa performance di retrieval (Recall@1 e @5) conferma che gli spazi latenti non si stanno allineando efficacemente.
